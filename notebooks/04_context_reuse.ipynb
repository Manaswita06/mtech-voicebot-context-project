```
from pathlib import Path
import json
report_path = Path('../outputs/context_reuse/context_reuse_report.json')
print('Exists:', report_path.exists())
if report_path.exists():
    report = json.loads(report_path.read_text())
    print('Session 1 extracted context:')
    print(json.dumps(report['session_1']['extracted_context'], indent=2))
    print('\nSession 2 replies:')
    print('Without context:', report['session_2']['without_context_reply'])
    print('With context:', report['session_2']['with_context_reply'])
```
